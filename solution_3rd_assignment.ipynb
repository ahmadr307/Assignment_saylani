**Q1:** Aik `Student` class banayein jismein `name` aur `roll_number` private variables (`__`) hon. In variables ko access aur modify karne ke liye getter aur setter methods likhein.

```python
# Q1: Student class with private variables and getter/setter methods

class Student:
    def __init__(self, name, roll_number):
        self.__name = name
        self.__roll_number = roll_number

    # Getter methods
    def get_name(self):
        return self.__name

    def get_roll_number(self):
        return self.__roll_number

    # Setter methods
    def set_name(self, name):
        self.__name = name

    def set_roll_number(self, roll_number):
        self.__roll_number = roll_number

# Test
s = Student("Ahmad", 101)
print("Name:", s.get_name())
print("Roll No:", s.get_roll_number())

s.set_name("Usama")
s.set_roll_number(102)
print("Updated Name:", s.get_name())
print("Updated Roll No:", s.get_roll_number())
```

**Q2:** Aik `BankAccount` class banayein jismein `__balance` private variable ho. Aik `deposit()` aur aik `withdraw()` method banayein, check karein ke withdraw karte waqt balance zero se kam na ho.

```python
# Q2: BankAccount class with deposit and withdraw methods

class BankAccount:
    def __init__(self, balance=0):
        self.__balance = balance

    def deposit(self, amount):
        self.__balance += amount
        print(f"Deposited: {amount} | Balance: {self.__balance}")

    def withdraw(self, amount):
        if self.__balance - amount < 0:
            print("Insufficient balance! Withdrawal not allowed.")
        else:
            self.__balance -= amount
            print(f"Withdrawn: {amount} | Balance: {self.__balance}")

    def get_balance(self):
        return self.__balance

# Test
acc = BankAccount(1000)
acc.deposit(500)
acc.withdraw(200)
acc.withdraw(2000)  # Should show error
```

**Q3:** Aik class banayein `Employee` jismein salary private ho. Agar koi salary ko minus (negative) value set karne ki koshish kare, toh setter method 'Invalid Salary' ka error message show kare.

```python
# Q3: Employee class with salary validation

class Employee:
    def __init__(self, name, salary):
        self.__name = name
        self.set_salary(salary)

    def get_salary(self):
        return self.__salary

    def set_salary(self, salary):
        if salary < 0:
            print("Invalid Salary")
        else:
            self.__salary = salary
            print(f"Salary set to: {self.__salary}")

# Test
e = Employee("Ahmad", 50000)
e.set_salary(-1000)   # Invalid
e.set_salary(60000)   # Valid
```

**Q4:** Protected variable (`_`) aur Private variable (`__`) ke beech kya farq hota hai? Aik simple code ke zariye dono ka use-case dikhayein.

```python
# Q4: Protected vs Private variables

class MyClass:
    def __init__(self):
        self._protected = "I am Protected"   # Single underscore
        self.__private = "I am Private"      # Double underscore

    def show(self):
        print(self._protected)
        print(self.__private)

class ChildClass(MyClass):
    def access(self):
        print(self._protected)    # Accessible in child class
        # print(self.__private)   # This would cause AttributeError

obj = MyClass()
obj.show()

child = ChildClass()
child.access()

# Protected can be accessed outside (but should not be, by convention)
print(obj._protected)

# Private cannot be accessed directly outside the class
try:
    print(obj.__private)
except AttributeError as e:
    print("Error:", e)
```

**Q5:** Encapsulation ka real-world faida kya hai? Private variable ko class ke bahar se direct access karne par kaunsa error aata hai? Code likh kar check karein.

```python
# Q5: Encapsulation benefit and private variable access error

class Person:
    def __init__(self, name, age):
        self.__name = name
        self.__age = age

    def get_info(self):
        return f"Name: {self.__name}, Age: {self.__age}"

p = Person("Ahmad", 20)
print(p.get_info())

# Trying to access private variable directly
try:
    print(p.__name)
except AttributeError as e:
    print("Error:", e)

# Benefit of Encapsulation:
# 1. Protects data from accidental modification
# 2. Controls how data is accessed or changed
# 3. Makes code more secure and maintainable
```

**Q7:** Multilevel Inheritance ka use karein. Class `Animal` -> Class `Mammal` -> Class `Dog`. Dog class ke object se parent classes ke features ko access karein.

```python
# Q7: Multilevel Inheritance - Animal -> Mammal -> Dog

class Animal:
    def __init__(self):
        self.alive = True

    def breathe(self):
        print("Animal breathes")

class Mammal(Animal):
    def __init__(self):
        super().__init__()
        self.warm_blooded = True

    def feed_milk(self):
        print("Mammal feeds milk to young ones")

class Dog(Mammal):
    def __init__(self, name):
        super().__init__()
        self.name = name

    def bark(self):
        print(f"{self.name} says: Woof!")

# Test
dog = Dog("Bruno")
dog.breathe()       # From Animal
dog.feed_milk()     # From Mammal
dog.bark()          # From Dog
print("Is alive:", dog.alive)
print("Warm blooded:", dog.warm_blooded)
```

**Q8:** Multiple Inheritance ka istemal karte hue do parent classes `Father` aur `Mother` banayein, aur unse aik child class `Child` derive karein jo dono ke habits inherit kare.

```python
# Q8: Multiple Inheritance - Father, Mother, Child

class Father:
    def habit_father(self):
        print("Father's habit: Reading books")

class Mother:
    def habit_mother(self):
        print("Mother's habit: Cooking")

class Child(Father, Mother):
    def habit_child(self):
        print("Child's habit: Playing games")

# Test
child = Child()
child.habit_father()   # Inherited from Father
child.habit_mother()   # Inherited from Mother
child.habit_child()    # Own habit
```

**Q9:** Child class ke constructor (`__init__`) ke andar parent class ke constructor ko call karne ke liye `super()` ka use kaise hota hai? Aik practical example likhein.

```python
# Q9: super() to call parent constructor

class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age
        print(f"Person created: {self.name}, Age: {self.age}")

class Student(Person):
    def __init__(self, name, age, roll_no):
        super().__init__(name, age)   # Calling parent constructor
        self.roll_no = roll_no
        print(f"Roll No: {self.roll_no}")

# Test
s = Student("Ahmad", 20, 101)
print(f"Name: {s.name}, Age: {s.age}, Roll: {s.roll_no}")
```

**Q10:** Hierarchical Inheritance kya hoti hai? Aik parent class `Shape` banayein aur us se do alag child classes `Circle` aur `Square` banakar samjhayein.

```python
# Q10: Hierarchical Inheritance - Shape -> Circle, Square

class Shape:
    def area(self):
        pass

    def describe(self):
        print("I am a Shape")

class Circle(Shape):
    def __init__(self, radius):
        self.radius = radius

    def area(self):
        return 3.14 * self.radius ** 2

class Square(Shape):
    def __init__(self, side):
        self.side = side

    def area(self):
        return self.side ** 2

# Test
c = Circle(5)
s = Square(4)

c.describe()
print("Circle Area:", c.area())

s.describe()
print("Square Area:", s.area())
```

**Q13:** Operator Overloading kya hoti hai? Do objects ko `+` operator ke zariye add karne ke liye `__add__` magic method ka use kaise karte hain? Code likhein.

```python
# Q13: Operator Overloading with __add__

class Point:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def __add__(self, other):
        return Point(self.x + other.x, self.y + other.y)

    def __str__(self):
        return f"Point({self.x}, {self.y})"

# Test
p1 = Point(2, 3)
p2 = Point(4, 5)
p3 = p1 + p2   # Calls __add__
print("P1:", p1)
print("P2:", p2)
print("P1 + P2 =", p3)
```

**Q15:** Kya Python default roop se Method Overloading (same method name with different arguments) support karta hai? Agar nahi, toh hum same name ke method ko multiple arguments ke sath kaise handle karte hain? Code likhein.

```python
# Q15: Method Overloading in Python

# Python does NOT support method overloading by default.
# If you define same method twice, the second one overwrites the first.

# Solution: Use default arguments or *args

class Calculator:
    def add(self, a, b=0, c=0):
        return a + b + c

calc = Calculator()
print("Add 2 numbers:", calc.add(5, 10))
print("Add 3 numbers:", calc.add(5, 10, 15))

# Using *args approach
class Calculator2:
    def add(self, *args):
        return sum(args)

calc2 = Calculator2()
print("Add with *args:", calc2.add(1, 2, 3, 4, 5))
```

**Q16:** `abc` module ka use karte hue aik abstract class `Computer` banayein jisme abstract method `process()` ho. Is class ka object banane ki koshish karein aur dekhein kya hota hai.

```python
# Q16: Abstract class using abc module

from abc import ABC, abstractmethod

class Computer(ABC):
    @abstractmethod
    def process(self):
        pass

# Try to create object of abstract class
try:
    c = Computer()
except TypeError as e:
    print("Error:", e)

# You cannot instantiate an abstract class directly!
```

**Q17:** Q16 wali `Computer` class ko inherit karke do child classes `Laptop` aur `Desktop` banayein aur dono mein `process()` method ko mandatory implement karein.

```python
# Q17: Child classes Laptop and Desktop from Computer

from abc import ABC, abstractmethod

class Computer(ABC):
    @abstractmethod
    def process(self):
        pass

class Laptop(Computer):
    def process(self):
        print("Laptop is processing using battery power")

class Desktop(Computer):
    def process(self):
        print("Desktop is processing using AC power")

# Test
laptop = Laptop()
laptop.process()

desktop = Desktop()
desktop.process()
```

**Q18:** Aik abstract class `Car` banayein jisme aik concrete (normal) method `start_engine()` ho aur aik abstract method `drive()` ho. Child class bana kar dono ko test karein.

```python
# Q18: Abstract class Car with concrete and abstract methods

from abc import ABC, abstractmethod

class Car(ABC):
    def start_engine(self):
        print("Engine started!")   # Concrete method

    @abstractmethod
    def drive(self):
        pass   # Abstract method

class ElectricCar(Car):
    def drive(self):
        print("Electric Car is driving silently")

class PetrolCar(Car):
    def drive(self):
        print("Petrol Car is driving with engine roar")

# Test
ec = ElectricCar()
ec.start_engine()   # Concrete method from parent
ec.drive()          # Overridden abstract method

pc = PetrolCar()
pc.start_engine()
pc.drive()
```

**Q19:** Abstraction aur Encapsulation ke darmiyan kya wazeh farq (main difference) hai? Dono ke concepts ko short statements mei define karein.

```python
# Q19: Difference between Abstraction and Encapsulation

# Abstraction:
# - Hides COMPLEXITY / implementation details
# - Shows only WHAT an object does, not HOW
# - Achieved using abstract classes and methods
# - Example: You press a button to turn on TV (you don't know internal circuit)

# Encapsulation:
# - Hides DATA / variables from outside access
# - Wraps data and methods together in a class
# - Achieved using private/protected variables
# - Example: ATM machine hides your PIN and bank logic

print("Abstraction  => Hides complexity, shows functionality")
print("Encapsulation => Hides data, protects it using access modifiers")

# Example of Abstraction
from abc import ABC, abstractmethod

class Shape(ABC):
    @abstractmethod
    def area(self):  # Only tells WHAT to do, not HOW
        pass

# Example of Encapsulation
class Person:
    def __init__(self, name):
        self.__name = name   # Hidden data

    def get_name(self):
        return self.__name
```

**Q20:** Aik real-world scenario (jaise Payment System) design karein jahan `Payment` aik abstract class ho aur `CreditCardPayment` aur `UPIPayment` uski subclasses hon.

```python
# Q20: Real-world Payment System using Abstract class

from abc import ABC, abstractmethod

class Payment(ABC):
    @abstractmethod
    def pay(self, amount):
        pass

    def payment_status(self):
        print("Payment Successful!")

class CreditCardPayment(Payment):
    def __init__(self, card_number):
        self.card_number = card_number

    def pay(self, amount):
        print(f"Paying Rs. {amount} using Credit Card ending in {self.card_number[-4:]}")
        self.payment_status()

class UPIPayment(Payment):
    def __init__(self, upi_id):
        self.upi_id = upi_id

    def pay(self, amount):
        print(f"Paying Rs. {amount} using UPI ID: {self.upi_id}")
        self.payment_status()

# Test
cc = CreditCardPayment("1234567890123456")
cc.pay(5000)

print()

upi = UPIPayment("ahmad@upi")
upi.pay(2000)
```
