### **Question 40**
(Visualization) Top 5 highest rated movies aur unke 'Meta_score' vs 'IMDB_Rating' (scaled to 100) ka ek Side-by-Side Comparison Bar Chart banayein.

```python
import numpy as np
top5 = data.nlargest(5, 'IMDB_Rating')[['Series_Title', 'IMDB_Rating', 'Meta_score']].reset_index(drop=True)
top5['IMDB_100'] = top5['IMDB_Rating'] * 10

x = np.arange(len(top5))
width = 0.35

plt.figure(figsize=(12, 6))
plt.bar(x - width/2, top5['IMDB_100'], width, label='IMDB (scaled to 100)', color='steelblue')
plt.bar(x + width/2, top5['Meta_score'], width, label='Meta Score', color='coral')
plt.xticks(x, top5['Series_Title'], rotation=20, ha='right')
plt.title('Top 5 Movies: IMDB Rating vs Meta Score')
plt.ylabel('Score')
plt.legend()
plt.tight_layout()
plt.show()
```
